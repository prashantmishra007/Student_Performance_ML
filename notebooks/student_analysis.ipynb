```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Dataset read karna
df = pd.read_csv('../dataset/StudentsPerformance.csv')

# 2. Starting rows display karna
print("--- Pehli 5 Rows ---")
display(df.head())

# 3. Columns data types aur info check karna
print("\n--- Dataset Summary & Types ---")
print(df.info())

# 4. Missing values check karna
print("\n--- Missing Values Count ---")
print(df.isnull().sum())
```

```python
import pandas as pd
import numpy as np

# 1. Dataset Load
df = pd.read_csv('../dataset/cleaned_students_performance.csv')

# 2. Features aur Target alag karna
X_df = df.drop(columns=['math_score', 'total_score', 'average_score'])
y = df['math_score'].values

# 3. Categorical variables ko One-Hot Encode karna (drop_first=True)
X_encoded = pd.get_dummies(X_df, drop_first=True)
feature_names = X_encoded.columns.tolist()
X = X_encoded.values.astype(float)

# 4. Train-Test Split (80% Train, 20% Test)
np.random.seed(42)
indices = np.arange(len(X))
np.random.shuffle(indices)

test_size = int(0.2 * len(X))
test_idx = indices[:test_size]
train_idx = indices[test_size:]

X_train, X_test = X[train_idx], X[test_idx]
y_train, y_test = y[train_idx], y[test_idx]

# 5. Multiple Linear Regression via Ordinary Least Squares (Normal Equation)
# Bias / Intercept term add karna
X_train_b = np.c_[np.ones((X_train.shape[0], 1)), X_train]
X_test_b = np.c_[np.ones((X_test.shape[0], 1)), X_test]

# Parameters calculate karna: theta = (X^T * X)^(-1) * X^T * y
theta = np.linalg.pinv(X_train_b.T @ X_train_b) @ X_train_b.T @ y_train

# 6. Evaluation on Test Data
y_pred = X_test_b @ theta

mae = np.mean(np.abs(y_test - y_pred))
rmse = np.sqrt(np.mean((y_test - y_pred) ** 2))
ss_res = np.sum((y_test - y_pred) ** 2)
ss_tot = np.sum((y_test - np.mean(y_test)) ** 2)
r2 = 1 - (ss_res / ss_tot)

print("--- Multiple Linear Regression Model Evaluation ---")
print(f"Total Test Samples: {len(y_test)}")
print(f"Mean Absolute Error (MAE) : {mae:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R-squared Score (R²)      : {r2:.4f}")

# Top 5 Sample Predictions Comparison
pred_comparison = pd.DataFrame({
    'Actual Math Score': y_test[:5],
    'Predicted Score': np.round(y_pred[:5], 2),
    'Difference (Error)': np.round(np.abs(y_test[:5] - y_pred[:5]), 2)
})
print("\n--- Sample Predictions ---")
display(pred_comparison)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Visualizing Predictions
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. Actual vs Predicted Scatter Plot
sns.scatterplot(x=y_test, y=y_pred, color='dodgerblue', alpha=0.7, ax=axes[0])
axes[0].plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
axes[0].set_title('Actual vs Predicted Math Scores', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Actual Math Score')
axes[0].set_ylabel('Predicted Math Score')

# 2. Residual Distribution (Errors)
residuals = y_test - y_pred
sns.histplot(residuals, kde=True, color='crimson', ax=axes[1])
axes[1].axvline(0, color='black', linestyle='--')
axes[1].set_title('Residuals (Error) Distribution', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Error (Actual - Predicted)')

plt.tight_layout()
plt.show()
```
